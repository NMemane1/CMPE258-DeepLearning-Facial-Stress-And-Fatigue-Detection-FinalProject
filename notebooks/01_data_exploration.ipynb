# 01 — Data Exploration
Quick EDA over the Drowsiness dataset + FER-2013, plus a peek at the class distribution and a few sample images. Runs anywhere (laptop, Colab, Kaggle).

```python
import sys
sys.path.insert(0, '..')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from src.data.dataset import DrowsinessDataset, FER2013Dataset, FER_EMOTION_TO_STRESS, FER_INT_TO_EMOTION, STRESS_CLASS_NAMES, FATIGUE_CLASS_NAMES
```

```python
# Adjust paths to where your data lives
DROWSY_PATH = '../data/processed/drowsiness'
FER_PATH    = '../data/processed/fer2013'

drowsy = DrowsinessDataset(DROWSY_PATH)
fer    = FER2013Dataset(FER_PATH)
print(f'Drowsiness samples: {len(drowsy)}')
print(f'FER-2013 samples:   {len(fer)}')
```

```python
# Class distribution
f_counts = Counter(r['fatigue_label'] for r in drowsy.records)
s_counts = Counter(r['stress_label']  for r in fer.records)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].bar([FATIGUE_CLASS_NAMES[k] for k in sorted(f_counts)], [f_counts[k] for k in sorted(f_counts)])
axes[0].set_title('Fatigue label distribution (Drowsiness)')
axes[0].set_ylabel('count')
axes[1].bar([STRESS_CLASS_NAMES[k] for k in sorted(s_counts)], [s_counts[k] for k in sorted(s_counts)])
axes[1].set_title('Stress label distribution (FER → stress remap)')
axes[1].set_ylabel('count')
plt.tight_layout()
plt.savefig('../assets/screenshots/class_distribution.png', dpi=120)
plt.show()
```

```python
# Sample images grid
fig, axes = plt.subplots(2, 6, figsize=(15, 5))
for i in range(6):
    rec = drowsy[i]
    axes[0, i].imshow(rec['image'])
    axes[0, i].set_title(f"fatigue={FATIGUE_CLASS_NAMES[rec['fatigue_label'].item()]}", fontsize=9)
    axes[0, i].axis('off')
for i in range(6):
    rec = fer[i*100]   # spread out across stress classes
    axes[1, i].imshow(rec['image'])
    axes[1, i].set_title(f"stress={STRESS_CLASS_NAMES[rec['stress_label'].item()]}", fontsize=9)
    axes[1, i].axis('off')
plt.tight_layout()
plt.savefig('../assets/screenshots/sample_grid.png', dpi=120)
plt.show()
```
